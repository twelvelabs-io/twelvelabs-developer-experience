# Authentication

All Jockey API requests require an API key passed in the `x-api-key` header. This notebook walks through obtaining, configuring, and validating your API key.

````python
import json
import os

import requests

# Configuration
API_KEY = os.environ.get("TWELVELABS_API_KEY", "YOUR_API_KEY")
BASE_URL = "https://api.twelvelabs.io/v1.3"
HEADERS = {"x-api-key": API_KEY, "Content-Type": "application/json"}
````

## Get Your API Key

1. Go to the [API Keys dashboard](https://playground.twelvelabs.io/dashboard/api-keys)
2. Create a new key or copy an existing one
3. Store it securely — treat it like a password

Set it as an environment variable before running this notebook:

```bash
export TWELVELABS_API_KEY="your-key-here"
```

## Header Format

Every request to the Jockey API must include the `x-api-key` header:

```
x-api-key: YOUR_API_KEY
```

## Python Example

Make an authenticated request to list your assets:

````python
response = requests.get(f"{BASE_URL}/assets", headers=HEADERS)

print(f"Status Code: {response.status_code}")
print(f"Response: {json.dumps(response.json(), indent=2)}")
````

## cURL Equivalent

The same request in cURL:

```bash
curl -H "x-api-key: YOUR_API_KEY" \
  https://api.twelvelabs.io/v1.3/assets
```

## JavaScript Equivalent

```javascript
const response = await fetch("https://api.twelvelabs.io/v1.3/assets", {
  headers: { "x-api-key": "YOUR_API_KEY" }
});
```

## Validate Your API Key

Use this helper function to confirm your key is valid before running other notebooks:

````python
def validate_api_key(api_key: str, base_url: str = BASE_URL) -> bool:
    """Validate an API key by making a lightweight request to the assets endpoint.

    Args:
        api_key: The Twelve Labs API key to validate.
        base_url: The base URL for the Jockey API.

    Returns:
        True if the key is valid, False otherwise.
    """
    headers = {"x-api-key": api_key}
    response = requests.get(f"{base_url}/assets", headers=headers)
    return response.status_code == 200


is_valid = validate_api_key(API_KEY)
print(f"API key valid: {is_valid}")
````

## Security Best Practices

- **Never hardcode API keys in source code** — use `os.environ.get()` as shown above
- **Use environment variables** — set `TWELVELABS_API_KEY` in your shell or `.env` file
- **Rotate keys periodically** from the [dashboard](https://playground.twelvelabs.io/dashboard/api-keys)
- **Use separate keys** for development and production environments

## Common Errors

| Error | Cause | Fix |
|-------|-------|-----|
| `401 Unauthorized` | Missing or invalid API key | Check your `x-api-key` header |
| `403 Forbidden` | Key doesn't have access | Verify key permissions in dashboard |

## Troubleshooting

Run this cell to diagnose authentication issues:

````python
def diagnose_auth(api_key: str, base_url: str = BASE_URL) -> None:
    """Diagnose common authentication issues.

    Args:
        api_key: The Twelve Labs API key to test.
        base_url: The base URL for the Jockey API.
    """
    if api_key == "YOUR_API_KEY":
        print("ERROR: API key is still the placeholder value.")
        print("Set TWELVELABS_API_KEY environment variable or replace the default.")
        return

    if not api_key:
        print("ERROR: API key is empty.")
        return

    headers = {"x-api-key": api_key}
    response = requests.get(f"{base_url}/assets", headers=headers)

    if response.status_code == 200:
        print("Authentication successful.")
    elif response.status_code == 401:
        print("ERROR: 401 Unauthorized — API key is invalid.")
        print("Check that the key matches your dashboard.")
    elif response.status_code == 403:
        print("ERROR: 403 Forbidden — key lacks required permissions.")
        print("Verify key permissions in the dashboard.")
    else:
        print(f"Unexpected status code: {response.status_code}")
        print(f"Response: {response.text}")


diagnose_auth(API_KEY)
````
